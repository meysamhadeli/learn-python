#### Match/Case (Python 3.10+)

A powerful switch-like statement

```python
def check_status(code):
    match code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500 | 502 | 503:  # Multiple values
            return "Server Error"
        case _:  # Default case
            return "Unknown"

print(f"Status 200: {check_status(200)}")
print(f"Status 404: {check_status(404)}")
print(f"Status 500: {check_status(500)}")
print(f"Status 418: {check_status(418)}")
```

```python
# Pattern matching with collections
def process_collection(data):
    match data:
        case []:
            return "Empty list"
        case [x]:
            return f"Single item: {x}"
        case [x, y]:
            return f"Two items: {x} and {y}"
        case {"name": name, "age": age}:
            return f"Person: {name}, {age} years old"
        case _:
            return "Unknown structure"

print(process_collection([]))
print(process_collection([42]))
print(process_collection([1, 2]))
print(process_collection({"name": "Alice", "age": 30}))
```

### Classes & Magic Methods

```python
class Person:
    # Class variable
    species = "Homo sapiens"
    
    # Constructor
    def __init__(self, name: str, age: int):
        self.name = name
        self._age = age
    
    # Regular method
    def greet(self) -> str:
        return f"Hi, I'm {self.name}"
    
    # Property with getter and setter
    @property
    def age(self):
        return self._age
    
    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value
    
    # Static method
    @staticmethod
    def is_adult(age: int) -> bool:
        return age >= 18
    
    # Class method
    @classmethod
    def from_birth_year(cls, name: str, year: int):
        age = 2026 - year
        return cls(name, age)

# Create instances
alice = Person("Alice", 30)
bob = Person.from_birth_year("Bob", 1990)

print(alice.greet())
print(bob.greet())
print(f"Alice's age: {alice.age}")
print(f"Is Alice adult? {Person.is_adult(alice.age)}")
print(f"Species: {Person.species}")
```

```python
# Magic methods
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Point({self.x}, {self.y})"
    
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

p1 = Point(1, 2)
p2 = Point(3, 4)
p3 = p1 + p2

print(f"p1: {p1}")
print(f"p2: {p2}")
print(f"p1 + p2 = {p3}")
print(f"p1 == Point(1, 2)? {p1 == Point(1, 2)}")
```

### Errors and Exceptions

#### Try-Except Blocks

```python
try:
    value = int(input("Enter a number: "))
    result = 100 / value
    print(f"Result: {result}")
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
except Exception as e:
    print(f"Unexpected error: {e}")
else:
    print("Success! No exceptions occurred.")
finally:
    print("This always runs (cleanup code here)")
```

#### Raising Exceptions

```python
def validate_age(age: int):
    if age < 0:
        raise ValueError("Age cannot be negative")
    if age > 150:
        raise ValueError("Age seems unrealistic")
    return age

try:
    validate_age(-5)
except ValueError as e:
    print(f"Error: {e}")

try:
    validate_age(200)
except ValueError as e:
    print(f"Error: {e}")
```

#### Custom Exceptions

```python
class PaymentError(Exception):
    """Custom exception for payment-related errors"""
    pass

def process_payment(amount):
    if amount <= 0:
        raise PaymentError(f"Invalid amount: {amount}. Amount must be positive.")
    if amount > 1000:
        raise PaymentError(f"Amount {amount} exceeds limit.")
    return f"Processed ${amount}"

try:
    print(process_payment(-50))
except PaymentError as e:
    print(f"Payment failed: {e}")

try:
    print(process_payment(1500))
except PaymentError as e:
    print(f"Payment failed: {e}")

print(process_payment(500))
```

### Generators

Generators produce values lazily, one at a time

```python
def fibonacci(limit: int):
    """Generate Fibonacci numbers up to limit"""
    a, b = 0, 1
    while a < limit:
        yield a
        a, b = b, a + b

# Using the generator
for num in fibonacci(100):
    print(num, end=" ")
print()
```

```python
# Generator expressions (lazy evaluation)
squares_gen = (x**2 for x in range(10))

# Values are generated on demand
print(f"First: {next(squares_gen)}")
print(f"Second: {next(squares_gen)}")
print(f"Third: {next(squares_gen)}")

# Convert to list if needed
remaining = list(squares_gen)
print(f"Remaining: {remaining}")
```

### Decorators

Decorators modify or enhance functions

```python
import time

def timer(func):
    """Decorator that measures function execution time"""
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        elapsed = time.time() - start
        print(f"{func.__name__} took {elapsed:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    """A function that takes some time"""
    time.sleep(1)
    return sum(range(1000000))

result = slow_function()
print(f"Result: {result}")
```

```python
# Decorator with arguments
def repeat(times):
    """Decorator that repeats a function multiple times"""
    def decorator(func):
        def wrapper(*args, **kwargs):
            results = []
            for _ in range(times):
                results.append(func(*args, **kwargs))
            return results
        return wrapper
    return decorator

@repeat(3)
def greet(name):
    return f"Hello, {name}"

print(greet("Alice"))
```

### Context Managers

Context managers handle setup and cleanup automatically

```python
# Using context manager with files (auto-closes)
with open("test.txt", "w") as f:
    f.write("Hello, World!")
# File is automatically closed here

# Reading the file
with open("test.txt", "r") as f:
    content = f.read()
    print(f"File content: {content}")
```

```python
# Custom context manager using class
class DatabaseConnection:
    def __enter__(self):
        print("Connecting to database...")
        self.connection = "fake_connection"
        return self.connection
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Closing database connection...")
        self.connection = None

with DatabaseConnection() as conn:
    print(f"Using {conn}")
```

```python
# Function-based context manager
from contextlib import contextmanager

@contextmanager
def timer(block_name):
    """Context manager that times a block of code"""
    print(f"Starting {block_name}...")
    start = time.time()
    try:
        yield
    finally:
        elapsed = time.time() - start
        print(f"{block_name} took {elapsed:.4f} seconds")

with timer("data processing"):
    sum(range(10_000_000))
    print("Processing complete")
```

## Chapter III: Concurrency

### Threading (for I/O-bound tasks)

```python
import threading
import time

def download(url: str):
    """Simulate downloading from a URL"""
    print(f"Starting download from {url}")
    time.sleep(1)  # Simulates I/O (GIL is released)
    print(f"Finished downloading from {url}")

# Create and start threads
threads = []
urls = ["a.com", "b.com", "c.com"]

for url in urls:
    t = threading.Thread(target=download, args=(url,))
    t.start()
    threads.append(t)

# Wait for all threads to complete
for t in threads:
    t.join()

print("All downloads complete!")
```

```python
# Using ThreadPoolExecutor (easier)
from concurrent.futures import ThreadPoolExecutor

def download(url):
    time.sleep(1)
    return f"Data from {url}"

urls = ["a.com", "b.com", "c.com", "d.com", "e.com"]

with ThreadPoolExecutor(max_workers=3) as executor:
    results = list(executor.map(download, urls))

for result in results:
    print(result)
```

### Async/Await

Single-threaded concurrency for high-performance I/O

```python
import asyncio

async def say_after(delay, msg):
    await asyncio.sleep(delay)
    print(msg)

async def main():
    # Run tasks concurrently
    task1 = asyncio.create_task(say_after(1, "Hello"))
    task2 = asyncio.create_task(say_after(2, "World"))
    
    await task1
    await task2

# Run the async main function
asyncio.run(main())
```

```python
# Running multiple tasks with gather
async def main():
    await asyncio.gather(
        say_after(1, "Hello"),
        say_after(2, "World"),
        say_after(1.5, "Python")
    )

asyncio.run(main())
```

### Multiprocessing (for CPU-bound work)

Use multiple processes to bypass the GIL and use multiple cores

```python
from multiprocessing import Pool

def cpu_intensive(n: int) -> int:
    """CPU-intensive calculation"""
    return sum(i * i for i in range(n))

if __name__ == "__main__":
    start = time.time()
    
    # Run on multiple processes
    with Pool(processes=4) as pool:
        results = pool.map(cpu_intensive, [5_000_000] * 4)
    
    elapsed = time.time() - start
    print(f"Parallel processing took {elapsed:.2f} seconds")
    print(f"Results: {results}")
```
